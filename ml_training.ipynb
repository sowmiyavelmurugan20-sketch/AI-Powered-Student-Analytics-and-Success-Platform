```python
import pandas as pd
m = pd.read_csv("students_master.csv")
print(m.shape)
print(m.risk_level.value_counts())
```

```python
m = m[m.student_status == "Active"]
drop = ["student_id","roll_no","student_name","photo_file","risk_score","risk_level",
        "top_challenge","recommended_action","dropout_risk_pct","true_journey_segment",
        "student_status","dropout_semester","mentor_id",
        "gender","area_type","first_gen_learner","family_income_band","scholarship"]
X = m.drop(columns=drop).select_dtypes("number")
y = m["risk_level"]
print(X.shape)
print(y.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import classification_report, confusion_matrix

model = make_pipeline(
    SimpleImputer(strategy="median"),
    RandomForestClassifier(300, class_weight="balanced", min_samples_leaf=2, random_state=0))

pred = cross_val_predict(model, X, y, cv=StratifiedKFold(5, shuffle=True, random_state=0))

print(classification_report(y, pred))
print(confusion_matrix(y, pred, labels=["Low","Moderate","High"]))
```

```python
import pandas as pd
model.fit(X, y)
imp = pd.Series(model[-1].feature_importances_, X.columns).sort_values(ascending=False)
print(imp.head(10).round(3))
```

```python
res = m[["student_name","roll_no","risk_level"]].copy()
res["predicted"] = model.predict(X)
proba = pd.DataFrame(model.predict_proba(X), columns=model.classes_, index=X.index)
res["P_high"] = proba["High"].round(2)
print(res.sort_values("P_high", ascending=False).head(10))
```

```python
import joblib
joblib.dump(model, "risk_model.pkl")
```

```python
hi = list(model.classes_).index("High")

def why(i, top=5):
    row = X.loc[[i]]
    base = model.predict_proba(row)[0, hi]
    drop_in_risk = {}
    for c in X.columns:
        t = row.copy(); t[c] = X[c].median()      # idha oru feature ah "typical student" value ku maathi paakkurom
        drop_in_risk[c] = base - model.predict_proba(t)[0, hi]
    return base, pd.Series(drop_in_risk).sort_values(ascending=False).head(top)

# highest-risk student
i = res.sort_values("P_high", ascending=False).index[0]
base, top = why(i)
print(m.loc[i, "student_name"], "P(High) =", round(base, 2))
print(top.round(3))
```
